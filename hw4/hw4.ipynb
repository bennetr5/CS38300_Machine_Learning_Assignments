# Part 1: Performance Measures for Evaluating a Binary Classifier  

```python
# Generate two arrays with 10 elements and each element is randomly either 0 or 1
# Variable "actual" array is referred to actual target or label
# Variable "predicted" array is referred to the prediction from a machine learningn model

import numpy as np
import matplotlib.pyplot as plt
import sklearn

np.random.seed(42)

# Create an array of 10 random integers between 0 and 1
actual = np.random.randint(0, 2, 10)

# Create another array of 10 random integers between 0 and 1
predicted = np.random.randint(0, 2, 10)

# Print the two arrays
print("Actual:", actual)
print("Predicted:", predicted)
```

```python
# TODO: calculate accuracy
accuracy = sklearn.metrics.accuracy_score(actual, predicted)
print(f"Accuracy: {accuracy}")
```

```python
# TODO: show confusion matrices
confusion_matrix = sklearn.metrics.confusion_matrix(actual, predicted)
display_confusion_matrix = sklearn.metrics.ConfusionMatrixDisplay(confusion_matrix)
display_confusion_matrix.plot()
```

```python
# TODO: show precision, recall, and F1 score
precision = sklearn.metrics.precision_score(actual, predicted)
recall = sklearn.metrics.recall_score(actual, predicted)
f1 = sklearn.metrics.f1_score(actual, predicted)
print(f"Precision: {precision}\nRecall: {recall}\nF1 Score: {f1}")
```

```python
# TODO: show TNR, FPR, FNR, and TPR
tp, tn, fp, fn = 0, 3, 4, 3
tnr = tn / (tn + fp)
fpr = 1 - tnr
fnr = fn / (fn + tp)
tpr = 1 - fnr
print(f"TNR: {tnr}\nFPR: {fpr}\nFNR: {fnr}\nTPR: {tpr}")
```

```python
# TODO: show the precision vs. recall plot
plt.figure()
plt.scatter(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.show()
```

```python
# TODO: show the ROC curve plot
fpr, tpr, thresholds = sklearn.metrics.roc_curve(actual, predicted)

plt.figure()
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], linestyle='--')  # Random classifier line
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```

```python
# TODO: Calcuate ROC AUC score
roc_auc = sklearn.metrics.roc_auc_score(actual, predicted)

print(f"ROC AUC Score: {roc_auc}")
```

# Part 2: Performance Measures for Evaluating a multiclass Classifier

```python
# Generate two arrays with 10 elements and each element is randomly between 0 or 4
# Variable "actual" array is referred to actual target or label
# Variable "predicted" array is referred to the prediction from a machine learningn model

import numpy as np
import sklearn
np.random.seed(42)

# Create an array of 10 random integers between 0 and 4
actual = np.random.randint(0, 5, 10)

# Create another array of 10 random integers between 0 and 4
predicted = np.random.randint(0, 5, 10)

# Print the two arrays
print("Actual:", actual)
print("Predicted:", predicted)
```

```python
# TODO: show confusion matrix
cm = sklearn.metrics.confusion_matrix(actual, predicted)
display_cm = sklearn.metrics.ConfusionMatrixDisplay(cm)
display_cm.plot()
```

```python
# TODO: show confusion matrix with errors only and errors normalized by row
cm_errors_only = sklearn.metrics.confusion_matrix(actual, predicted)
np.fill_diagonal(cm_errors_only, 0)

display_cm_errors = sklearn.metrics.ConfusionMatrixDisplay(cm_errors_only)
display_cm_errors.plot()
plt.title("Errors Only")

cm_normalized = sklearn.metrics.confusion_matrix(actual, predicted, normalize='true')
display_cm_normalized = sklearn.metrics.ConfusionMatrixDisplay(cm_normalized)
display_cm_normalized.plot()
plt.title("Normalized")
```
